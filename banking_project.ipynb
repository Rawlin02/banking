```python
#converting numerical data into catogrical data (estimated income column)
bins = [0,100000, 300000,float('inf')]
labels = ['low','mid','high']
df['income band'] = pd.cut(df['Estimated_Income'], bins=bins, labels=labels,right=False)
```

```python
df['income band'].value_counts().plot(kind='bar')
```

```python
# examine the distribution of unique categories in categorical columns
categorical_cols = df[["BRId","IAId", "GenderId", "Amount_of_Credit_Cards", "Nationality", "Occupation","Fee_Structure", "Loyalty_Classification", "Properties_Owned", "Risk_Weighting"]].columns
for col in categorical_cols:
    print(f"value counts for '{col}':")
    display(df[col].value_counts())
```

```python
# univariate Analysis
cols=["BRId","IAId", "GenderId", "Amount_of_Credit_Cards", "Nationality", "Occupation","Fee_Structure", "Loyalty_Classification", "Properties_Owned", "Risk_Weighting"]
df["GenderId"]= df["GenderId"].astype(str)
for i, predictor in enumerate(cols):
    plt.figure(i)
    df[predictor]= df[predictor].astype(str)
    plt.figure(figsize=(8,4))
    sns.countplot(data=df,x=predictor, hue='GenderId')
    plt.title(f'countplot for {predictor}')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
for i, predictor in enumerate(df[["BRId","IAId", "GenderId", "Amount_of_Credit_Cards", "Nationality", "Occupation","Fee_Structure", "Loyalty_Classification", "Properties_Owned", "Risk_Weighting"]]):
    plt.figure(i)
    sns.countplot(data=df, x=predictor, hue='GenderId')
```

```python
# histplot of value counts for different occupation
for col in categorical_cols:
    if col == "Occupation":
        continue
    plt.figure(figsize=(8,4))
    sns.histplot(df[col])
    plt.title('histogram of occupation counnt')
    plt.xlabel(col)
    plt.ylabel("count")
    plt.show()
```
